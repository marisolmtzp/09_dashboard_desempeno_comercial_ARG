# 📊 Proyecto: Dashboard de desempeño comercial 2024–2025

La empresa **Andes Retail Group** es una empresa de retail con operaciones en **Perú, Chile y Colombia** que comercializa productos en **cuatro categorías: Electrónica, Ropa, Deportes y Hogar**.

La **dirección ejecutiva** necesita un **dashboard interactivo** que permita entender el **desempeño comercial de los años 2024–2025**. Actualmente, la información está dispersa en datos transaccionales y ** no existe una visión clara que permita responder preguntas estratégicas sobre ventas, rentabilidad y comportamiento de clientes**. El objetivo es transformar estos datos desordenados en un dashboard que presente la información visual, clara y accionable.

Este notebook servirá como guía estructurada para el desarrollo del proyecto.

### Vista Overview - ¿Cómo ha evolucionado el ingreso total entre 2024 y 2025?
Se incluirán los KPIs:
- 1. **Ingresos Totales.** La métrica de mayor jerarquía en cualquier negocio de retail porque representa el punto de partida de toda la rentabilidad y el indicador definitivo de la demanda del mercado.
- 2. **Ganancia Total.** Por que es la métrica de salud financiera que indica la rentabilidad real del negocio (el dinero que realmente se queda la empresa) después de deducir el costo de los bienes vendidos.
- 3. **Pedidos Totales.** Esta métrica descompone el problema en sus partes. Ingresos = Pedidos × Ticket promedio (Por sí solo responde: "¿estamos perdiendo clientes/transacciones, o son los mismos clientes comprando menos?")
- 4. **Ticket Promedio** Se lee junto al KPI de Pedidos Totales para explicar la disminución de ingresos mostrando que la caída no tiene una sola causa dominante, sino que es una combinación de menos transacciones y menor valor por transacción.

A continuación se desglosa qué gráfico se usará para cada pregunta y la justificación de la elección:
- **Gráfico de tiempo:** Gráfico lineal - **Evolución Mensual 2024 vs 2025** (gráfico lineal mensual con comparativa entre los dos años). Para responder a la pregunta **¿Cómo ha evolucionado el ingreso total entre 2024 y 2025?**.<br>
Muestra en una sola vista la comparativa de los montos en el mismo periodo para los dos años, haciendo evidente aumentos y caídas.<br>Este gráfico también responde a la pregunta **¿Qué patrones temporales se observan a lo largo del año?** ya que muestra claramente las dinámicas de estacionalidad, ventas altas a finales y principios de año comunes en retail.
- **Gráfico comparación cliente:** Responde a la pregunta **¿Qué segmentos de clientes aportan mayor ingreso y rentabilidad?**, al filtrar por país, se contempla la dimensión geográfica.
- **Gráfico comparación productos:** Responde a la pregunta - **¿Qué categorías de producto tienen mayor impacto en el negocio?** , al filtrar por país, se contempla la dimensión geográfica.<br>

Descripción de la jerarquía visual (Vista General):
- **Título:** Desempeño Comercial 2024 - 2025
